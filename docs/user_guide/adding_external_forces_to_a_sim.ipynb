## Conservative Forces and Potentials
🚧 *Still working on it...* 🚧

Conservative forces and their associated potentials expressed as a potential can be added to a simulation with the {meth}`~tambora.simulation.Sim.add_external_pot` method. tambora currently only accepts galpy potentials.

### galpy potentials

{meth}`~tambora.simulation.Sim.add_external_pot` is a convenience method that creates a {class}`~tambora.dynamics.forces.external_force.ExternalGalpyPotential` instance for the provided galpy potential and adds it to the simulation.

Ensure that the provided potential has physical units turned on. If not turned on, the provided potential is ambigous and an internal scaling of ro=8 kpc, vo =220 km/s will be assumed.


````python
from galpy.potential import NFWPotential
extpot = NFWPotential()
extpot.turn_physical_on()

from tambora.simulation import Sim
mysim = Sim()
mysim.add_external_pot(extpot)
````
