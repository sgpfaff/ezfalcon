# Unit Conventions

tambora used two sets of units: one that is physically motivated for all inputs and outputs that the user interacts with and another that is derived from a [fundamental set of units](#fundamental-units) for all operations under the hood. 

## Fundamental Units
| Quantity | Internal Unit | User I/0 Unit |
| -------- | -------- | -------- |
| Length | $\text{kpc}$ | $\text{kpc}$ |
| Mass | $\text{M}_{\odot}$| $\text{M}_{\odot}$ | 
| Time | $\text{Gyr}$ | $\text{Gyr}$ |

## Velocity and Acceleration
| Quantity | Internal Unit | User I/0 Unit |
| -------- | -------- | -------- |
| Velocity | $\text{kpc} / \text{Gyr}$ | $\text{km} / \text{s}$ |
| Angular Velocity | $ 1 / \text{Gyr}$ | $\text{km} / \text{s} / \text{kpc}$ |
| Angle | $\text{rad}$ | $\text{rad}$ | 
| Acceleration | $\text{kpc} / \text{Gyr}^2$ | $\text{km} / \text{s} / \text{Gyr}$|

## Energy And Momentum
| Quantity | Internal Unit | User I/0 Unit |
| -------- | -------- | -------- |
| Momentum | $\text{M}_{\odot} \text{ kpc } / \text{ Gyr}$ | $\text{M}_{\odot} \text{ km } / \text{ s}$ |
| Angular Momentum | $\text{M}_{\odot} \text{ kpc}^2 / \text{ Gyr}$ | $\text{M}_{\odot} \text{ km}^2 / \text{ s}$ |
| Energy (Potential, Kinetic, Total) | $\text{M}_{\odot} \text{ kpc}^2 / \text{Gyr}^2 $ | $\text{M}_{\odot} \text{ km}^2 / \text{s}^2 $|


## Returning Method Outputs in Internal Units

One may return outputs of {class}`~tambora.simulation.Sim` methods in tambora internal units by passing `return_internal=True` as an argument.
