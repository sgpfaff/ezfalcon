# Overview of Force Classes

The acceleration a particle feels can depend on many factors, including:
* **its own** position, velocity, and/or mass
* the positions, velocities, and/or mass of **other particles**
* the **time** it is evaluated

Which of these factors determines how a particle accelerates depends on the force that produces it. Different forces can require various combinations of these properties depending on the physical mechanism it encodes. Further, some forces possess additional mathematical properties of interest (namely the potential of conservative forces). The forces in tambora can be thought of as being organized in the following way:

|          | Conservative (depends only on position) | Non-Conservative (depends not only on position)| 
| -------- | ------------ | ---------------- |
| **External** (independent of other particles) | galpy potentials  | galpy dynamical friction |
| **Interaction** (depends on other particles)| self-gravity  | collisional forces, pressure, etc. *Not Included.*  |

**External forces** are those that are constructed and act on particles individually. That is to say, external forces act on particles with no knowledge of the other particles. Broadly speaking, these forces encode interaction mechanisms between particles and their (often idealized) environment.

Meanwhile, **interaction forces** encode interaction mechanisms *between* particles and thus depend on the state of the ensemble of particles at a given time. 

The methods to find the acceleration for these two types of forces differ dramatically. Because of this, tambora provides two base classes for external forces ({class}`~tambora.dynamics.forces.ExternalForce`) and interaction forces ({class}`~tambora.dynamics.forces.InteractionForce`). 

All forces have an `acc` method.

**Conservative** forces have the unique property that they can be implicitly defined with a potential. Forces are identified as conservative in tambora with the {class}`~tambora.dynamics.forces.Conservative` mixin that provides additional methods for the potential.

In some cases, it is more efficient to calculate the acceleration and potential together (e.g., in direct summation). Thus, there is also an optional `acc_and_potential` method for conservative forces to provide an efficient way to calculate both acceleration and potential if available. By default, `acc_and_potential` calculates the acceleration and potential separately and returns them together. Forces/Force solvers that *can* calculate both together faster than separately override the default with its own `acc_and_potential` method.

## Custom Forces
🚧 *Still working on it...* 🚧

If you'd like to add a custom force to your simulation or want an easy way to combine it with other forces in tambora, you can create a new subclass of one of the existing force classes or create your own for your force. The input and output of the methods must be in [tambora internal units](units.ipynb#units-conventions) for it to be combined with other forces and integrated into a simulation properly.
