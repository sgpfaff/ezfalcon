# Force Classes

Forces are represented in a standardized format as descendents of {class}`~tambora.dynamics.forces.BaseForce`, which has a single abstract acceleration method {meth}`~tambora.dynamics.forces.BaseForce.acc`. 

By default, the acceleration depends on the position, velocity, and mass of the particles and the time it is evaluated. However, not all forces depend on all of these quantities and some possess additional properties (such as the potential for conservatives forces, as discussed below).

## Conservative Forces ({class}`~tambora.dynamics.forces.ConservativeForce`)
🚧 *Still working on it...* 🚧

Conservative forces are unique in that: (i) their acceleration field depends only on position (and possibly time as well), and (ii) they can be represented with a potential. These properties of conservative forces inform how conservative forces are represented in tambora.

Conservative forces in tambora are descendants of {class}`~tambora.dynamics.forces.ConservativeForce`, which is a subclass of {class}`~tambora.dynamics.forces.BaseForce`. {class}`~tambora.dynamics.forces.ConservativeForce` overrides the acceleration method of {class}`~tambora.dynamics.forces.BaseForce` with it's own {meth}`~tambora.dynamics.forces.ConservativeForce.acc` abstract method that depends only on position, mass, and time. Mass is included as an argument because it is needed for calculating self-gravity, but a particle's own mass is not used in calculating the accleration or potential it feels from the conservative force.

{class}`~tambora.dynamics.forces.ConservativeForce` has an additional abstract {meth}`~tambora.dynamics.forces.ConservativeForce.potential` method. In some cases, it is more efficient to calculate the acceleration and potential together (e.g., in direct summation). Thus, there is also a {meth}`~tambora.dynamics.forces.ConservativeForce.acc_and_potential` method that provides an efficient way to calculate both acceleration and potential when possible. By default, {meth}`~tambora.dynamics.forces.ConservativeForce.acc_and_potential` calculates the acceleration and potential separately and returns them together. Forces/Force solvers that *can* calculate both together faster than separately override the default with its own `acc_and_potential` method.

## Custom Forces
🚧 *Still working on it...* 🚧

If you'd like to add a custom force to your simulation or want an easy way to combine it with other forces in tambora, you can create a new subclass of one of the existing force classes or create your own for your force. The input and output of the methods must be in [tambora internal units](units.ipynb#units-conventions) for it to be combined with other forces and integrated into a simulation properly.
