## Setting up the Simulation
*More details about setting up a simulation can be found in the [user-guide section dedicated to this topic](user_guide/setting_up_a_simulation.ipynb#setting-up-a-simulation).*

#### Creating the Simulation

Simulations are instances of the {class}`~tambora.simulation.Sim` class, which are simple to instantiate:

```python
from tambora.simulation import Sim

mysim = Sim()
```

You will quickly become familiar with the {class}`~tambora.simulation.Sim` class, as most of the features of tambora run through it.
